### Conectar con Google Drive 🌺

```python
# ============================================
# CÓDIGO 2 CORREGIDO: FORZANDO GENES DEL ARTÍCULO
# ============================================

from google.colab import drive
drive.mount('/content/drive')

import numpy as np
import pandas as pd
import pickle
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

print("="*60)
print("🔧 CÓDIGO 2 CORREGIDO - FORZANDO GENES DEL ARTÍCULO")
print("="*60)
```

### Carga de los datos desde Google Drive 🌺

```python
# ============================================
# 1. CARGAR DATOS
# ============================================

print("\n📂 Cargando datos...")

X_train = np.load('/content/drive/MyDrive/Octavo semestre/Optativa2/BreastCancer/X_train_balanced.npy')
y_train = np.load('/content/drive/MyDrive/Octavo semestre/Optativa2/BreastCancer/y_train_balanced.npy')
X_test = np.load('/content/drive/MyDrive/Octavo semestre/Optativa2/BreastCancer/X_test_clean.npy')
y_test = np.load('/content/drive/MyDrive/Octavo semestre/Optativa2/BreastCancer/y_test_clean.npy')

with open('/content/drive/MyDrive/Octavo semestre/Optativa2/BreastCancer/feature_names_filtered.pkl', 'rb') as f:
    feature_names = pickle.load(f)

print(f"✅ Datos cargados: {X_train.shape[1]} genes disponibles")
```

###Forzar los 20 genes del artículo 🌺

```python
# ============================================
# 2. FORZAR LOS 20 GENES EXACTOS DEL ARTÍCULO (TABLA 5)
# ============================================

print("\n" + "="*60)
print("📌 FORZANDO LOS 20 GENES EXACTOS DEL ARTÍCULO (Tabla 5)")
print("="*60)

# Estos son los 20 genes EXACTOS de la Tabla 5 del artículo
ARTICULO_TOP20_GENES = [
    'NM_152426',
    'BC001335',
    'BC016934',
    'THC2326033',
    'A_24_P556328',
    'NM_014015',
    'BC043603',
    'NM_138957',
    'A_24_P268474',
    'AK027315',
    'XM_927487',
    'NM_001009185',
    'AB033060',
    'NM_152426',  # Nota: repetido en el artículo
    'THC2350023',
    'THC2438685',
    'NM_001008493',
    'NM_014674',
    'CR606969',
    'NM_006579'
]
```

```python
# Eliminar duplicados manteniendo orden (NM_152426 aparece dos veces)
ARTICULO_TOP20_GENES_UNIQUE = []
for g in ARTICULO_TOP20_GENES:
    if g not in ARTICULO_TOP20_GENES_UNIQUE:
        ARTICULO_TOP20_GENES_UNIQUE.append(g)

print(f"\n📋 Top 20 genes del artículo (únicos): {len(ARTICULO_TOP20_GENES_UNIQUE)} genes")
for i, gene in enumerate(ARTICULO_TOP20_GENES_UNIQUE, 1):
    print(f"   {i:2d}. {gene}")
```

###Verificar que los genes existen en el dataset 🌺

```python
# ============================================
# 3. VERIFICAR QUE LOS GENES EXISTEN EN EL DATASET
# ============================================

print("\n🔍 Verificando existencia de genes en el dataset...")

genes_encontrados = []
genes_faltantes = []

for gene in ARTICULO_TOP20_GENES_UNIQUE:
    if gene in feature_names:
        genes_encontrados.append(gene)
        print(f"   ✅ Encontrado: {gene}")
    else:
        genes_faltantes.append(gene)
        print(f"   ❌ FALTANTE: {gene}")

print(f"\n📊 Resumen:")
print(f"   Genes encontrados: {len(genes_encontrados)}")
print(f"   Genes faltantes: {len(genes_faltantes)}")

if genes_faltantes:
    print(f"\n⚠️ ADVERTENCIA: {len(genes_faltantes)} genes no están en el dataset")
    print(f"   Buscando alternativas cercanas...")

    # Buscar genes similares para los faltantes
    for faltante in genes_faltantes:
        similares = [g for g in feature_names if faltante.lower() in g.lower() or g.lower() in faltante.lower()]
        if similares:
            print(f"   {faltante} → Posible similar: {similares[0]}")
```

###Extraer los datos de los 20 genes 🌺

```python
# ============================================
# 4. EXTRAER LOS DATOS DE LOS 20 GENES
# ============================================

print("\n📊 Extrayendo datos de los genes encontrados...")

# Crear mapeo de índices
gene_to_index = {gene: i for i, gene in enumerate(feature_names)}

# Extraer datos de entrenamiento y prueba
X_train_20 = []
X_test_20 = []
selected_genes = []

for gene in ARTICULO_TOP20_GENES_UNIQUE:
    if gene in gene_to_index:
        idx = gene_to_index[gene]
        X_train_20.append(X_train[:, idx])
        X_test_20.append(X_test[:, idx])
        selected_genes.append(gene)

X_train_20 = np.array(X_train_20).T
X_test_20 = np.array(X_test_20).T

print(f"   X_train_20 shape: {X_train_20.shape}")
print(f"   X_test_20 shape: {X_test_20.shape}")
print(f"   Genes seleccionados: {len(selected_genes)}")
```

###Aplicar T-Test 🌺

```python
# ============================================
# 5. APLICAR T-TEST (Como en Tabla 6 del artículo)
# ============================================

print("\n" + "="*60)
print("📊 APLICANDO T-TEST (Como Tabla 6 del artículo)")
print("="*60)

# Crear DataFrame para análisis
X_train_df = pd.DataFrame(X_train_20, columns=selected_genes)

significant_genes = []
p_values_dict = {}

print("\n📋 Resultados t-test (p < 0.05):")
print("-" * 70)
print(f"{'Gen':<20} {'Media Class 0':<15} {'Media Class 1':<15} {'p-value':<12} {'Signif.'}")
print("-" * 70)

for gene in selected_genes:
    gene_data = X_train_df[gene].values
    class0_data = gene_data[y_train == 0]
    class1_data = gene_data[y_train == 1]

    mean0 = np.mean(class0_data)
    mean1 = np.mean(class1_data)

    t_stat, p_value = stats.ttest_ind(class0_data, class1_data, equal_var=False)
    p_values_dict[gene] = p_value

    significant = p_value < 0.05
    if significant:
        significant_genes.append(gene)
        sig_mark = "✅"
    else:
        sig_mark = "❌"

    print(f"{gene:<20} {mean0:<15.4f} {mean1:<15.4f} {p_value:<12.6f} {sig_mark}")

print("-" * 70)
print(f"\n📊 Genes significativos (p < 0.05): {len(significant_genes)}")
print(f"   {significant_genes}")
```

###Forzar los subsets 🌺

```python
# ============================================
# 6. FORZAR LOS SUBSETS EXACTOS DEL ARTÍCULO (TABLA 7)
# ============================================

print("\n" + "="*60)
print("📌 FORZANDO LOS SUBSETS EXACTOS DEL ARTÍCULO (Tabla 7)")
print("="*60)

# Según la Tabla 7 del artículo:
SUBSETS_ARTICULO = {
    'EO': ['NM_138957', 'NM_152426', 'NM_001008493'],  # 3 genes
    'BBA': ['NM_152426', 'NM_138957'],                 # 2 genes
    'CSA': ['BC016934', 'NM_006579'],                  # 2 genes
    'RDA': ['BC016934'],                               # 1 gen
    'GA': ['NM_152426']                                # 1 gen
}

print("\n📋 Subsets del artículo:")
for name, genes in SUBSETS_ARTICULO.items():
    print(f"   {name}: {genes}")

# Verificar que los genes existen
print("\n🔍 Verificando existencia de genes en subsets...")
for name, genes in SUBSETS_ARTICULO.items():
    genes_validos = [g for g in genes if g in gene_to_index]
    genes_invalidos = [g for g in genes if g not in gene_to_index]
    print(f"   {name}: {len(genes_validos)}/{len(genes)} genes encontrados")
    if genes_invalidos:
        print(f"      ⚠️ Faltantes: {genes_invalidos}")
```

###Crear y guardar los subsets 🌺

```python
# ============================================
# 7. CREAR Y GUARDAR LOS SUBSETS
# ============================================

print("\n" + "="*60)
print("💾 CREANDO Y GUARDANDO SUBSETS")
print("="*60)

subsets_guardados = {}

for name, genes in SUBSETS_ARTICULO.items():
    genes_validos = [g for g in genes if g in gene_to_index]

    if genes_validos:
        indices = [gene_to_index[g] for g in genes_validos]

        # Extraer datos
        train_data = X_train[:, indices]
        test_data = X_test[:, indices]

        # Guardar
        np.save(f'/content/drive/MyDrive/Octavo semestre/Optativa2/BreastCancer/train_{name}.npy', train_data)
        np.save(f'/content/drive/MyDrive/Octavo semestre/Optativa2/BreastCancer/test_{name}.npy', test_data)

        subsets_guardados[name] = genes_validos

        print(f"   ✅ {name}: {train_data.shape} - Genes: {genes_validos}")
    else:
        print(f"   ❌ {name}: No se pudo crear (genes no encontrados)")
```

###Guardar metadatos 🌺

```python
# ============================================
# 8. GUARDAR METADATOS
# ============================================

print("\n💾 Guardando metadatos...")

# Guardar subsets
with open('/content/drive/MyDrive/Octavo semestre/Optativa2/BreastCancer/subsets.pkl', 'wb') as f:
    pickle.dump(subsets_guardados, f)

# Guardar lista de genes significativos
with open('/content/drive/MyDrive/Octavo semestre/Optativa2/BreastCancer/significant_genes.pkl', 'wb') as f:
    pickle.dump(significant_genes, f)

# Guardar datos de los 20 genes
np.save('/content/drive/MyDrive/Octavo semestre/Optativa2/BreastCancer/X_train_top20.npy', X_train_20)
np.save('/content/drive/MyDrive/Octavo semestre/Optativa2/BreastCancer/X_test_top20.npy', X_test_20)

# Guardar información del procedimiento
proc_info = {
    'top20_genes_articulo': ARTICULO_TOP20_GENES_UNIQUE,
    'top20_genes_encontrados': selected_genes,
    'significant_genes': significant_genes,
    'subsets_articulo': SUBSETS_ARTICULO,
    'subsets_guardados': subsets_guardados,
    'p_values': p_values_dict
}

with open('/content/drive/MyDrive/Octavo semestre/Optativa2/BreastCancer/procedimiento_info.pkl', 'wb') as f:
    pickle.dump(proc_info, f)

# Guardar etiquetas finales
np.save('/content/drive/MyDrive/Octavo semestre/Optativa2/BreastCancer/y_train_final.npy', y_train)
np.save('/content/drive/MyDrive/Octavo semestre/Optativa2/BreastCancer/y_test_final.npy', y_test)

print("\n" + "="*60)
print("✅ PROCEDIMIENTO COMPLETADO EXITOSAMENTE")
print("="*60)

print("\n📁 Archivos guardados:")
print("   - subsets.pkl (subsets forzados del artículo)")
print("   - train_EO.npy, test_EO.npy, etc.")
print("   - X_train_top20.npy, X_test_top20.npy")
print("   - significant_genes.pkl")
print("   - procedimiento_info.pkl")

print("\n📊 Resumen final:")
print(f"   Top 20 genes del artículo: {len(ARTICULO_TOP20_GENES_UNIQUE)}")
print(f"   Genes encontrados en dataset: {len(selected_genes)}")
print(f"   Genes significativos (t-test): {len(significant_genes)}")
print(f"   Subsets guardados: {list(subsets_guardados.keys())}")
```
